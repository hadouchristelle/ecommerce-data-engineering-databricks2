```python
transactions_path = (
    "/Volumes/real_time_catalogue/"
    "bronze_realtime/"
    "landing_events/"
    "transactions"
)

dbutils.fs.mkdirs(transactions_path)

print("Dossier créé :", transactions_path)
```

```python
import json
import random
import time
import uuid
from datetime import datetime, timezone


def generate_transaction():
    quantity = random.randint(1, 5)
    unit_price = round(random.uniform(10, 500), 2)

    return {
        "event_type": "transaction",
        "transaction_id": f"RT-{uuid.uuid4().hex[:12].upper()}",
        "customer_id": f"C{random.randint(1, 10000):05d}",
        "product_id": f"P{random.randint(1, 1000):04d}",
        "quantity": quantity,
        "unit_price": unit_price,
        "total_amount": round(quantity * unit_price, 2),
        "transaction_date": datetime.now(timezone.utc).isoformat(),
        "source_type": "realtime"
    }


number_of_events = 10

for event_number in range(1, number_of_events + 1):
    transaction = generate_transaction()

    file_name = (
        f"{transactions_path}/"
        f"transaction_{int(time.time() * 1000)}_"
        f"{event_number}.json"
    )

    dbutils.fs.put(
        file_name,
        json.dumps(transaction),
        overwrite=False
    )

    print(f"Événement {event_number} créé")
    print(json.dumps(transaction, indent=2))
    print("-" * 60)

    time.sleep(2)

print("Génération terminée.")
```
