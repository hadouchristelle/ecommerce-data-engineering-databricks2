```python
# Lire le fichier customers depuis ADLS
df_customers = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("abfss://commerce@adlgenstorage.dfs.core.windows.net/bronze/customers_raw_imperfect.csv")
)

display(df_customers)
```

```python
# Lire le fichier products depuis ADLS
df_products = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("abfss://commerce@adlgenstorage.dfs.core.windows.net/bronze/products_raw_imperfect.csv")
)

display(df_products)
```

```python
# Lire le fichier sessions depuis ADLS
df_sessions = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("abfss://commerce@adlgenstorage.dfs.core.windows.net/bronze/sessions_raw_imperfect.csv")
)

display(df_sessions)
```

```python
# Lire le fichier transactions depuis ADLS
df_transactions = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("abfss://commerce@adlgenstorage.dfs.core.windows.net/bronze/transactions_raw_imperfect.csv")
)

display(df_transactions)
```

```python
# Enregistrer customers en table Delta Bronze
df_customers.write.format("delta").mode("overwrite").saveAsTable("`E-commerce`.bronze.customers")
```
