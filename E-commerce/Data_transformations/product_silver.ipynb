```python
# Lire la table products Bronze
df_products = spark.table("`E-commerce`.bronze.products")
```

```python
# Afficher les types des colonnes
df_products.printSchema()
```

```python
# Supprimer les lignes identiques
products_silver = df_products.dropDuplicates()
```

```python
# Garder un seul produit par product_id
products_silver = products_silver.dropDuplicates(["product_id"])
```

```python
from pyspark.sql.functions import col# Supprimer les lignes sans product_id

products_silver = products_silver.filter(
    col("product_id").isNotNull()
)
```

```python
# Supprimer les lignes sans product_id
products_silver = products_silver.filter(
    col("product_id").isNotNull()
)
```

```python
# Supprimer les espaces inutiles
from pyspark.sql.functions import trim

for c, dtype in products_silver.dtypes:
    if dtype == "string":
        products_silver = products_silver.withColumn(c, trim(col(c)))
```

```python
# Comparer Bronze et Silver
print("Bronze :", df_products.count())
print("Silver :", products_silver.count())
```

```python
# Enregistrer products dans Silver
products_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("`E-commerce`.silver.products")
```
