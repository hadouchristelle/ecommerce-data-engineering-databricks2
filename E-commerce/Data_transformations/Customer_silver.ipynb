```python
from pyspark.sql.functions import *
from pyspark.sql.functions import trim
```

```python
# Lire la table customers Bronze
df_customers = spark.table("`E-commerce`.bronze.customers")
```

```python
# Afficher les types des colonnes
df_customers.printSchema()
```

```python
# Supprimer les lignes identiques
customers_silver = df_customers.dropDuplicates()
```

```python
# Garder un seul enregistrement par customer_id
customers_silver = customers_silver.dropDuplicates(["customer_id"])
```

```python
 col# Supprimer les lignes sans customer_id

customers_silver = customers_silver.filter(
    col("customer_id").isNotNull()
)
```

```python
# Supprimer les espaces inutiles dans les textes

for c, dtype in customers_silver.dtypes:
    if dtype == "string":
        customers_silver = customers_silver.withColumn(c, trim(col(c)))
```

```python
# Afficher customers après nettoyage
display(customers_silver)
```

```python
# Enregistrer customers dans Silver
customers_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("`E-commerce`.silver.customers")
```
