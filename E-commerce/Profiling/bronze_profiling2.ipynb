```python
# Compter les lignes et les colonnes de chaque table
tables = ["customers", "products", "sessions", "transactions", "reviews"]

for table in tables:
    df = spark.table(f"`E-commerce`.bronze.{table}")
    print(table, ":", df.count(), "lignes,", len(df.columns), "colonnes")
```

```python
# Afficher le schema de chaque table
for table in tables:
    print("\nTABLE :", table)
    spark.table(f"`E-commerce`.bronze.{table}").printSchema()
```

```python
# Compter les valeurs NULL dans chaque colonne
from pyspark.sql.functions import col, sum

for table in tables:
    print(f"\nNULL dans {table}:")
    
    df = spark.table(f"`E-commerce`.bronze.{table}")
    
    df.select([
        sum(col(c).isNull().cast("int")).alias(c)
        for c in df.columns
    ]).show()
```

```python
# Compter les lignes en double dans chaque table
for table in tables:
    df = spark.table(f"`E-commerce`.bronze.{table}")
    
    doublons = df.count() - df.dropDuplicates().count()
    
    print(f"{table}: {doublons} doublons")
```
